```python
# If your Saudade v3 files are attached as a Kaggle input dataset, copy them into the working dir.
# Adjust SRC_DIR to wherever your dataset is mounted.
import shutil, os

SRC_DIR = "/kaggle/input/saudade-v3"   # <-- change to your dataset path
DST_DIR = "/kaggle/working"

if os.path.isdir(SRC_DIR):
    for fname in ["config.py", "model.py", "train_tokenizer.py", "train.py", "generate.py", "data.txt"]:
        src = os.path.join(SRC_DIR, fname)
        if os.path.exists(src):
            shutil.copy(src, os.path.join(DST_DIR, fname))
    print("Copied Saudade v3 files into", DST_DIR)
else:
    print(f"SRC_DIR {SRC_DIR} not found — make sure your dataset is attached, or upload files directly.")

os.chdir(DST_DIR)
```
